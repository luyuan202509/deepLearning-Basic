## 三层网络

### 从输入层到第 1 层的信号传递

```python
X = np.array([1,0.5])
W1 = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
B1 = np.array([0.1, 0.2, 0.3])

print(W1.shape)
print(X.shape)
print(B1.shape)

A1 = np.dot(X,W1) +B1 
print(A1)
```

```python
Z1 = sigmoid(A1)
print("激活前的值:",A1)
print("激活后的值Z1:",Z1)
```

### 第一层到第二层

```python
W2= np.array([[0.1, 0.4], [0.2, 0.5], [0.3, 0.6]])
B2 = np.array([0.1, 0.2])
print(Z1.shape) # (3,)
print(W2.shape) # (3, 2) 
print(B2.shape) # (2,)

A2 = np.dot(Z1, W2) + B2
Z2 = sigmoid(A2)
print(Z2)
```

### 第二层到第三层

```python
W3 = np.array([[0.1, 0.3], [0.2, 0.4]])
B3 = np.array([0.1, 0.2])
A3 = np.dot(Z2,W3) + B3
Z3 = sigmoid(A3)
print(Z3)
```
